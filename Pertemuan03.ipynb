```python
nilai_mahasiswa = int(input("85: "))
standar_kelulusan = int(input("77: "))

apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print(f"85: {nilai_mahasiswa}")
print(f"77: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
nilai_mahasiswa1 = int(input("85: "))
nilai_mahasiswa2 = int(input("70: "))
nilai_mahasiswa3 = int(input("90: "))
nilai_mahasiswa4 = int(input("60: "))
nilai_mahasiswa5 = int(input("80: "))

standar_kelulusan = int(input("73: "))

nilai_mahasiswa = [
    nilai_mahasiswa1,
    nilai_mahasiswa2,
    nilai_mahasiswa3,
    nilai_mahasiswa4,
    nilai_mahasiswa5
]

for i, nilai in enumerate(nilai_mahasiswa, start=1):
    apakah_lulus = nilai >= standar_kelulusan
    perlu_remedial = nilai < standar_kelulusan

    print(f"\nMahasiswa {i}")
    print(f"85, 70, 90, 60, 80: {nilai}")
    print(f"73: {standar_kelulusan}")
    print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
    print(f"Perlu Remedial? {perlu_remedial}")
```
